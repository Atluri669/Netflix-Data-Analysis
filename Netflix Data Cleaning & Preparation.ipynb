```python
# STEP 1: IMPORT THE DATASET
import pandas as pd

df = pd.read_csv("Dataset.csv")
```

```python
# STEP 2: UNDERSTAND THE DATASET
print("\nDataset shape (rows, columns):")
print(df.shape)

print("\nDataset names:")
print(df.columns.tolist())

print("\nDatasetinfo:")
print(df.info())

print("\nStatistical summary:")
print(df.describe(include='all').T)
```

```python
# STEP 3: CHECK MISSING VALUES

print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nMissing values percentage in each column:")
missing_percentage = (df.isnull().sum() * 100).round(2)
print(missing_percentage[missing_percentage > 0])
```

```python
# STEP 4: REMOVE DUPLICATE RECORDS

print("\nDuplicate rows before cleaning:")
print(df.duplicated().sum())

# Remove completely identical rows
df = df.drop_duplicates()

# Remove duplicate show IDs if the column exists
if "showid" in df.columns:
    df = df.drop_duplicates(subset=["Show_id"], keep="first")

print("Duplicate records removed.")
print("Dataset shape after removing duplicates:",df.shape)
```

```python
# STEP 5: CLEAN TEXT AND FORMATTING INCONSISTENCIES

# Clean extra spaces and standardize text in column names
df.columns = df.columns.str.strip().str.lower()



# Clean extra spaces in text columns

for column in df.select_dtypes(include=["object","string"]).columns:
    df[column] = df[column].astype(str).str.strip()
# Convert blank strings into missing values

df = df.replace(r"^\s*$",pd.NA, regex=True)

print("\nText formatting cleaned.")
```

```python
# STEP 6: STANDARDIZE TYPE, COUNTRY, AND RATING

# Standardize Type

if "type" in df.columns:
    df["type"] = df["type"].str.title()

# Standardize Rating
if "rating" in df.columns:
    df["rating"] = df["rating"].str.upper()

# Standardize Country

if"country" in df.columns:
    df["country"] = df["country"].str.replace(
        r"\s*\s*",",",regex = True
    )
    
print("\nType, Rating, and Country standardized.")
```

```python
# STEP 7: HANDLE MISSING VALUES

# Fill missing text fields with "Unknown"
text_columns = [
    "director",
    "cat",
    "country",
    "rating",
    "duration",
    "listed_in",
    "description"
]

for column in text_columns:
    if column in df.columns:
        df[column] = df[column].fillna("Unknown")
# Convert release_year to numeric if the column exists

if "release_year" in df.columns:
    df["release_year"]= pd.to_numeric(df["release_year"], errors="coerce")


# Convert date_added to datetime
if "data_added" in df.columns:
    df["date_added"] = pd.to_datetime(
        df["date_added"], errors="coerce"
    )
    
print("\nMissing values handled.")
print(df.isnull().sum())
```

```python
# STEP 8: CHECK DATA TYPES AND DATA QUALITY

print("\nData types after cleaning:")
print(df.dtypes)

if "type" in df.columns:
    print("nUnique content types:")
    print(df["type"].value_counts(dropna=False))
    
if "rating" in df.columns:
    print("\nUnique ratings:")
    print(df["rating"].value_counts(dropna=False))

if "release_year" in df.columns:
    print("\nRelease year summary:")
    print(df["release_year"].describe())
```

```python
# STEP 9: EXPORT THE CLEANED DATASET

output_file = "Netflix_cleaned.csv"

df.to_csv(output_file, index=False)

print("\nCleaning completed Succesfully!")
print("Cleaned dataset saved as:", output_file)
print("Final dataset shape:", df.shape)
print("Final duplicate row count:",df.duplicated().sum())
```
